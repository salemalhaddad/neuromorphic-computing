There are generations of neural networks:
1. Perceptrons: mostly revolved around boolean outputs
2. Activation function: here, continuous outputs are introduced. Majority of AI work is here.
3. Spiking neuron: Another stage receiving momumentum recently with a focus on synaptic, low-power computing.

The main reason behing the emergence of SNNs is how current ANNs (specifically in the next gen) are not perfectly simulating the power efficient / spike-firing model an actual neuron follows. An example is illustrated in the paper Maas, 1996:
> a single cortical area involved in visual processing can complete its computation in just 20-30 msec (Rolls, 1994; Rolls & Tovee, 1994). On the other hand, the firing rates of neurons involved in these computations are usually below 100Hz, and hence at least 20 - 30 msec would be needed just to sample the current firing rate of a neuron. Thus a coding of analog variables by firing rates seems quite dubious in the context of fastcortical computations

In other words, rate coding is less fast than maybe other methods. A particular fast method is the usage of timing of single action potentials (i.e. spikes).

Enter SNNs: a specific neural network architecture that uses spiking neurons as its core component. What's novel about this new form of neural networks is how it '[uses] time as a resource for computation and communication.'

In a SNN, a neuron v fires whenever its potential $P_{v}$ reaches a certain threshold $\Theta_{v}$, and $P_{v}$ (potential of neuron v) = EPSPs + IPSPs, with these two resulting from firing other 'presynaptic' neurons u connected via synapse to neuron v.

> The firing of a "presynaptic" neuron u at time s contributes to the potential $P_{v}$ at time t an amount that is modelled by the term $w_{u,v}*e(t - s)$ which consists of a weight $w_{u,v}$ and a response function $e(t - s)$. Note: $w_{u,v} \ge 0$

The threshold function $\Theta_{v}$ embeds the following two into its logic:
1) A neuron won't fire for some time t ms, no matter the input's strength
2) A neuron will only fire after t ms passes, and only if the input is strong enough

A SNN also takes into consideration when that neuron fires. Mathematially, if the presynaptic neuron's set of firing times is $F_{u}$, then the potential at the trigger zone (part of neuron where spikes are initiated) of neuron v at time t is:

$P_v(t) := \sum_{u: <u,v> \in E} \sum_{s \in F_u: s < t} w_{u,v} \cdot \varepsilon_{u,v}(t - s)$
