There are three generations of neural networks:
1. Perceptrons: mostly revolved around boolean outputs
2. Activation function: here, continuous outputs are introduced. Majority of AI work is here.
3. Spiking neuron: Another stage gaining momentum recently with a focus on synaptic, low-power computing.

The main reason behing the emergence of SNNs is how current ANNs (specifically in the next gen) are not perfectly simulating the power efficient / spike-firing model an actual neuron follows. An example is illustrated in the paper Maas, 1996:
> a single cortical area involved in visual processing can complete its computation in just 20-30 msec (Rolls, 1994; Rolls & Tovee, 1994). On the other hand, the firing rates of neurons involved in these computations are usually below 100Hz, and hence at least 20 - 30 msec would be needed just to sample the current firing rate of a neuron. Thus a coding of analog variables by firing rates seems quite dubious in the context of fastcortical computations

In other words, rate coding is less fast than maybe other methods. A particular fast method is the usage of timing of single action potentials (i.e. spikes).

Enter SNNs: a specific neural network architecture that uses spiking neurons as its core component. What's novel about this new form of neural networks is how it '[uses] time as a resource for computation and communication.'

In a SNN, a neuron v fires whenever its potential $P_{v}$ reaches a certain threshold $\Theta_{v}$, and $P_{v}$ (potential of neuron v) = EPSPs + IPSPs, with these two resulting from firing other 'presynaptic' neurons u connected via synapse to neuron v.

> The firing of a "presynaptic" neuron u at time s contributes to the potential $P_{v}$ at time t an amount that is modelled by the term $w_{u,v}*e(t - s)$ which consists of a weight $w_{u,v}$ and a response function $e(t - s)$. Note: $w_{u,v} \ge 0$

The threshold function $\Theta_{v}$ embeds the following two into its logic:
1) A neuron won't fire for some time t ms, no matter the input's strength
2) A neuron will only fire after t ms passes, and only if the input is strong enough


![Threshold function](threshold-function.png)

A SNN also takes into consideration when that neuron fires. Mathematially, if the presynaptic neuron's set of firing times is $F_{u}$, then the potential at the trigger zone (part of neuron where spikes are initiated) of neuron v at time t is:

$P_v(t) := \sum_{u: <u,v> \in E} \sum_{s \in F_u: s < t} w_{u,v} \cdot \varepsilon_{u,v}(t - s)$

where:
- $V$: finite set of **spiking neurons**
- $E \subseteq V\times V$: set of **synapses** (directed connections); $\langle u,v\rangle\in E$ means a synapse from neuron $u$ to neuron $v$
- $u,v \in V$: neurons (presynaptic $u$, postsynaptic $v$)
- $t \in \mathbb{R}^+$: time (nonnegative real)
- $\mathbb{R}^+ := \{x\in\mathbb{R}: x\ge 0\}$: set of nonnegative real numbers
- $F_u \subseteq \mathbb{R}^+$: set of **firing times** (spike times) of neuron $u$
- $s \in F_u$: a spike time of neuron $u$
- $s<t$: only spikes occurring **before** time $t$ contribute
- $w_{u,v}\ge 0$: **synaptic weight** of synapse $\langle u,v\rangle$
- $\varepsilon_{u,v}:\mathbb{R}^+\to\mathbb{R}$: **response function** (postsynaptic potential kernel) of synapse $\langle u,v\rangle$
- $\varepsilon_{u,v}(t-s)$: contribution at time $t$ from a spike at time $s$, depending on elapsed time $t-s$
- $P_v(t)$: **potential** at the trigger zone of neuron $v$ at time $t$


SNNs with Boolean Input and Output
- These neuron models receive boolean (bit vectors) as input and output bits too
- Coincidence detection (or pattern matching) function is a function that outputs 1 if two distinct input vectors have at least one value (at a particular index i) in common (equal)
![Coincidence-detection function](coincidence-detection-function.png)
- SNNs of Type A, i.e. of the deterministic type, can computer coincidence functions when delays to the neuron v is equal between both the first bit vector and the second vector ($\Delta_{a_i, v} = \Delta_{b_i, v}$)
- For such a SNN to be 'noise-robust', it should assign the spiking neuron's threshold as 1.5 so that if there's one spike (1 < 1.5), nothing happens, and if two spikes coincide, firing happens (2 > 1.5). Anything between 0 and 2 won't fire.
- This is powerful since a threshold or sigmoidal neural net would take at minimum n/log(n+1) and $\sqrt(n)$, respectively.

SNNs with Analog Input and Boolean Output
- These SNNs take in real numbers as input unlike boolean-input SNNs
- To encode a real number inputs, they're represented as spikes and only fired at $T_{input}-x_{i}*c$ where $T_{input}$ depends on the time when input arrives while the output can be either 1) boolean, similar to the previous SNN, or 2) output spikes at $T_{output}-y*c$ where output time > input time 
- For SNNs with analog input and boolean output, where input is positive real numbers and output is boolean, element distinct function is used:
![Element Distinct Function](element-distinct-function.png)
- Noise robustness is achieved by adding a condition for when $\min{(\lvert x_i - x_j \rvert : i \neq j)}$ has a value between 0 and 1 for some input vector x (of real numbers) - this avoids any noise-inducing/"aribtrary" outputs that affect overall model
- The resulting element distinct function is superior to both layered threshold circuit and a sigmoidal neural net as theorem 2 and 3 suggests:

Conclusion
- SNNs are at least of the same compute power as neural nets from first two generations (layers MLPs and sigmoidal NNs), with a single spiking neuron sometimes being equivalent to a multi-layered MLP or a sigmoid neural net
- Noise robustness is critical in SNNs for it to perform accurately through the usage of functions defined above
- Theorems and their proofs in this paper show the complexity superiority of a single spiking neuron
- This paper popularized the SNN and introduced mathematical proofs for the case of SNNs > ANNs
